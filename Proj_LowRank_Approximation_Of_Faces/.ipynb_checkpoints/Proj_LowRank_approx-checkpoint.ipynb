## 1. Load and Format the Data

Each of our face images (127x127 grayscale) had already been flattened into a single
row and saved to faces.csv, with one row per image. We load this CSV into a NumPy
matrix A, where each row corresponds to one midshipman's photo and each column
corresponds to a pixel position across all images.

```python
from PIL import Image
from pathlib import Path
import numpy as np
import os
import matplotlib.pyplot as plt
import pandas as pd
```

Conversion: CSV -> dataframe -> matrix

```python
df = pd.read_csv("faces.csv")
A = df.to_numpy()
```

## 2. Singular Value Decomposition

We compute the SVD of A, decomposing it into U, S (singular values), and Vt.
This decomposition lets us later approximate A using only its most significant
values, discarding lower-significance components that may represent noise
rather than meaningful data.

```python
#SVD
U, S, Vt = np.linalg.svd(A, full_matrices=False)
```

## 3. Rank and Singular Value Plot

We check the rank of A and plot its singular values on a log scale to see how
quickly they drop off. This plot is what influenced our initial choice of `k=127` as a
first cutoff, since values beyond that point appeared to drop off rapidly and
contribute relatively little.

```python
print(f"Number of singular values: {len(S)}")
rank = np.linalg.matrix_rank(A)
print(f"Rank: {rank}")

plt.figure()
plt.plot(S, marker="o")
plt.xlabel("Index")
plt.ylabel("Singular value")
plt.title("Singular values of A")
plt.yscale("log")
plt.show()
```

## 4. Low-Rank Matrix Reconstruction

Here we trim U, S, and Vt down to their first k components and reassemble a
low-rank approximation A prime of the original data. We varied k across several values
(1, 5, 127, 500, 1500) to compare how reconstruction quality changes as more or
fewer components are kept.  
We use the two helper functions defined below to do this for different k's.  
The first helper reconstructs A' given U, S, Vt, and a k.  
The second helper displays the original compared to the reconstructed images given A, A', a test set of images, k, and dimensions.

```python
def reconstruct(U, S, Vt, k):
    U_p = U[:, :k]
    S_p = S[:k]
    Vt_p = Vt[:k, :]
    A_p = (U_p * S_p) @ Vt_p

    return A_p
```

```python
def compare_reconstructions(A, A_p, indices, k, shape=(127, 127)):
    images_to_show = []
    titles = []
    for idx in indices:
        images_to_show.append(A[idx])
        images_to_show.append(A_p[idx])
        titles.append(f"Original {idx}")
        titles.append(f"k={k} {idx}")

    fig, axes = plt.subplots(len(indices), 2, figsize=(8, 4 * len(indices)))

    for ax, img, title in zip(axes.flat, images_to_show, titles):
        ax.imshow(img.reshape(shape), cmap="gray")
        ax.axis("off")
        ax.set_title(title, fontsize=10)

    plt.tight_layout()
    plt.show()
```

## 5. Comparing Original vs. Low-Rank Reconstructions

We display the original image alongside its rank-k reconstruction for each of our
5 test photos, repeating this comparison across every k value tested. This is the
data referenced in the Results Analysis section below, showing the
progression from heavily degraded (low k) to visually similar (high k)
reconstructions.

```python
#manually selected images to display diversity in testing
test_indices = [1, 53, 12, 20, 57]
```

```python
k = 127
A_p = reconstruct(U, S, Vt, k)
compare_reconstructions(A, A_p, test_indices, k)
```

```python
k = 500
A_p = reconstruct(U, S, Vt, k)
compare_reconstructions(A, A_p, test_indices, k)
```

```python
k = 1500
A_p = reconstruct(U, S, Vt, k)
compare_reconstructions(A, A_p, test_indices, k)
```

```python
k = 5
A_p = reconstruct(U, S, Vt, k)
compare_reconstructions(A, A_p, test_indices, k)
```

```python
k = 1
A_p = reconstruct(U, S, Vt, k)
compare_reconstructions(A, A_p, test_indices, k)
```

## 6. Results Analysis

    Each is represented in a 2x5 table format of images, where the left hand column is the original and the right hand column is the SVD-modified low rank approximation. We used the same 5 pictures for each change in k, that way results could be compared more accurately. Each of the 5 pictures represented a different type of midshipman as we used male and female pictures, glasses, and different ethnicities. 

	With our initial k=127, the resulting pictures after SVD were somewhat similar to the originals, but a couple could have been tougher to identify the original person(the darker colored photos perhaps due to less contrast). We then decided to increase the k value and find a good upper limit balance. How far could we actually decrease rank without getting too low and too blurred of a photo result? Increasing to k=500, we still had some blur but the pictures were much clearer. Further going to 1500 almost made the pictures perfect. Going higher didn’t yield as much benefit for each increase in k. 

	Then, we decided to test the lower end of the spectrum. Reducing k further below 127 rapidly decreased the image quality. At k=5, we noticed an interesting point in that all pictures started to resemble each other, but still had bits of the originals making each one unique. However, the interesting result comes when we reduce rank all the way down to the principal value at k=1, representing the feature found in the most amount of midshipmen. At k=1, all the SVD-modified pictures were the same, regardless of the original’s race, gender, or other features. One may say this represents the quintessential midshipman.
